```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../../data/processed/sales/customer_segments_analysis.csv").set_index(keys=["rank"])
df
```

```python
_, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 6))

sns.barplot(
    data=df,
    x="customer_segment",
    y="revenue",
    hue="customer_segment",
    ax=ax1,
)
ax1.set(
	title="Revenue by Customer Segment",
	xlabel="Customer Segment",
	ylabel="Revenue ($)",
)

sns.barplot(
    data=df,
    x="customer_segment",
    y="profit",
    hue="customer_segment",
    ax=ax2,
)
ax2.set(
	title="Profit by Customer Segment",
	xlabel="Customer Segment",
	ylabel="Profit ($)",
)

for ax in (ax1, ax2):
	ax.set_ylim(0, 3.6e06)
	ax.grid(axis="y", linestyle="--", linewidth=0.5)

plt.suptitle("Revenue and Profit by Customer Segment", fontsize=16)
plt.tight_layout()
plt.show()
```
