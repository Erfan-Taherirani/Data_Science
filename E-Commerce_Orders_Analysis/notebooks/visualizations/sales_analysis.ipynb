```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../../data/processed/sales/customer_segments_analysis.csv").set_index(keys=["rank"])
df
```

```python
_, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 6))

sns.barplot(
    data=df,
    x="customer_segment",
    y="revenue",
    hue="customer_segment",
    ax=ax1,
)
ax1.set(
	title="Revenue by Customer Segment",
	xlabel="Customer Segment",
	ylabel="Revenue ($)",
)

sns.barplot(
    data=df,
    x="customer_segment",
    y="profit",
    hue="customer_segment",
    ax=ax2,
)
ax2.set(
	title="Profit by Customer Segment",
	xlabel="Customer Segment",
	ylabel="Profit ($)",
)

for ax in (ax1, ax2):
	ax.set_ylim(0, 3.6e06)
	ax.grid(axis="y", linestyle="--", linewidth=0.3)

plt.suptitle("Revenue and Profit by Customer Segment", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
df = pd.read_csv("../../data/processed/sales/sales_channel_analysis.csv")
df
```

```python
_, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 6))

sns.barplot(
    data=df,
    x="sales_channel",
    y="revenue",
    hue="sales_channel",
    ax=ax1
)
sns.barplot(
    data=df,
    x="sales_channel",
    y="profit",
    hue="sales_channel",
    ax=ax2
)

ax2.sharey(ax1)

plt.suptitle("Sales Channels by Revenue and Profit", fontsize=16)
ax1.set(
    title="Sales Channels by Revenue",
    xlabel="Sales Channel",
    ylabel="Revenue"
)
ax2.set(
    title="Sales Channels by Profit",
    xlabel="Sales Channel",
    ylabel="Profit"
)
for ax in (ax1, ax2):
	ax.grid(axis="y", alpha=0.3, linestyle="--")
	
plt.tight_layout()
plt.show()
```
