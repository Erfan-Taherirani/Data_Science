```python
import numpy as np
import pandas as pd
```

```python
def weekly_expenses(breakfast_expenses, lunch_expenses, dinner_expenses):
	food_expenses = breakfast_expenses + lunch_expenses + dinner_expenses
	total_expenses = food_expenses

	return total_expenses

breakfast_expenses = 60
lunch_expenses = (2 * 35) + (3 * 25)
dinner_expenses = (2 * 35) + (2 * 25)

food_expenses = weekly_expenses(breakfast_expenses, lunch_expenses, dinner_expenses)
```

```python
import datetime

now = datetime.datetime.now()
now.strftime("%Y-%m-%d")
```

```python
df = pd.DataFrame(
    {
        "date": now.date(),
        "weekly_food_expenses": food_expenses,
        "weekly_transportation_expenses": "N/A",
        "weekly_housing_expenses": "N/A",
        "weekly_clothing_expenses": "N/A",
        "weekly_other_expenses": "N/A",
        "weekly_cafeteria_expenses": "N/A",
        "weekly_total_expenses": "N/A",
        
	},
    index=[0],
    dtype=object
)
df
```
